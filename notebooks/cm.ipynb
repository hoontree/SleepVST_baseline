```python
import numpy as np
import matplotlib.pyplot as plt

cm = np.array([
    [0.73, 0.20, 0.03, 0.04],
    [0.16, 0.53, 0.23, 0.07],
    [0.15, 0.24, 0.57, 0.04],
    [0.12, 0.29, 0.05, 0.54]
])

counts = np.array([
    [10412, 2813, 440, 507],
    [5131, 16637, 7295, 2139],
    [817, 1321, 3093, 219],
    [820, 2080, 340, 3854]
])

classes = ["Wake", "Light", "Deep", "REM"]

precision = [0.61, 0.73, 0.28, 0.57]
recall = [0.73, 0.53, 0.57, 0.54]

fig, ax = plt.subplots(figsize=(6,6))

# spacing이 있는 confusion matrix
mesh = ax.pcolormesh(
    cm,
    cmap="Blues",
    vmin=0,
    vmax=1,
    edgecolors="white",   # 셀 간격
    linewidth=1           # 간격 두께
)

# ticks
ax.set_xticks(np.arange(4) + 0.5)
ax.set_yticks(np.arange(4) + 0.5)

ax.set_xticklabels(classes, fontsize=16)
ax.set_yticklabels(classes, fontsize=16, rotation=90, va="center")

ax.tick_params(
    axis="both",
    which="both",
    length=4,   # tick 길이
    width=1
)

ax.set_xlabel("Predicted", fontsize=16)
ax.set_ylabel("True", fontsize=16)

ax.invert_yaxis()

# 셀 텍스트
for i in range(4):
    for j in range(4):

        # cm 값 (큰 폰트)
        ax.text(
            j + 0.5,
            i + 0.42,
            f"{cm[i,j]:.2f}",
            ha="center",
            va="center",
            fontsize=18
        )

        # count 값 (작은 폰트 유지)
        ax.text(
            j + 0.5,
            i + 0.63,
            f"({counts[i,j]})",
            ha="center",
            va="center",
            fontsize=14
        )

# Precision (top)
for j, p in enumerate(precision):
    ax.text(j + 0.5, -0.05, f"{p:.2f}", ha="center", fontsize=16)

# Recall (right)
for i, r in enumerate(recall):
    ax.text(4, i + 0.5, f"{r:.2f}", va="center", fontsize=16)

# 제목
ax.text(2, -0.3, "Per-class Precision", ha="center", fontsize=18)

ax.text(4.5, 2, "Per-class Recall", rotation=90, va="center", fontsize=18)

# 테두리 제거
for spine in ax.spines.values():
    spine.set_visible(False)


plt.tight_layout()
plt.show()
```
