```python
def _get_last_row_column_value(file_path: str, column_name: str) -> str:
    """Get the last row value from a specific column in CSV file"""
    with open(file_path, 'r', newline='') as f:
        f.seek(0, 2)  # Move to end of file
        file_size = f.tell()

        buffer = bytearray()
        pointer = file_size - 1

        # Read line backwards
        while pointer >= 0:
            f.seek(pointer)
            byte = f.read(1)
            if byte == '\n' and buffer:
                break
            buffer.insert(0, ord(byte))
            pointer -= 1

        last_line = buffer.decode('utf-8')

        # Read header
        f.seek(0)
        reader = csv.reader(f)
        header = next(reader)
        col_index = header.index(column_name)

        # Parse last line
        last_values = list(csv.reader([last_line]))[0]
        return last_values[col_index]
```

```python
video_path= "/tf/00_data/#_2021_Sleep_Video/"  # Path to the input video directory
sum = 0
annotation_dir = "/tf/00_AIoT2/video_signal/#_2021_Sleep_Video/30sec_labels"
count = 0

motion_path = "data/motionfeatures"
for video_file in sorted(Path(video_path).glob("A*/*.mp4")):
    record_id = (video_file.name).replace("_video_01.mp4", "")
    ann_file = Path(annotation_dir) / f"{record_id}_label.csv"
    motion_file = Path(motion_path) / f"{record_id}_motion_features.npy"
    if not ann_file.exists():
        print(f"Annotation file not found for video: {video_file}")
        continue
    cap = cv2.VideoCapture(str(video_file))
    if not cap.isOpened():
        print(f"Error opening video file: {video_file}")
        continue

    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    target_fps= 4
    frame_interval= int(round(fps / target_fps))
    total_steps = max(1, int(frame_count // frame_interval) - 1)
    
    total_epochs = int(_get_last_row_column_value(str(ann_file), 'Start_Epoch'))
    

    matched = analysis[(analysis['record_id'] == record_id) & (analysis['split'] == 'test')]['motion_epochs']
    
    motion = np.load(motion_file, allow_pickle=True).item()
    
    if matched.empty:
        continue
    try:
        # motion_epochs = matched.iloc[0]
        motion_epochs = {k: len(v) // 150 for k, v in motion.items()}
        print(record_id, motion_epochs)
        # diff = total_steps / (fps * 30) - motion_epochs
        # print(f"{record_id}: {diff}")
        # sum += diff
        count += 1
    except Exception as e:
        print(f"Record ID : {record_id} - error: {e}")

    cap.release()
    
print("Total difference in epochs:", sum, count)
```
